```python
import pandas as pd
import numpy as np
from google.colab import files
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.preprocessing import image
from sklearn.metrics.pairwise import cosine_similarity
import os
import zipfile
import mimetypes
```

```python
print("📥 Upload your dataset file (CSV or Excel):")
uploaded = files.upload()
dataset_filename = list(uploaded.keys())[0]

mime = mimetypes.guess_type(dataset_filename)[0]

if dataset_filename.endswith(".xlsx") or dataset_filename.endswith(".xls"):
    df = pd.read_excel(dataset_filename)
else:
    try:
        df = pd.read_csv(dataset_filename)
    except UnicodeDecodeError:
        df = pd.read_csv(dataset_filename, encoding="ISO-8859-1")

df.columns = df.columns.str.strip()
print("✅ Dataset loaded successfully!")
print("Columns:", df.columns)
```

```python
print("\n📂 Upload a ZIP file of reference crop images:")
uploaded_folder = files.upload()
zip_filename = list(uploaded_folder.keys())[0]

with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall("crop_images")

print("✅ Reference images extracted.")
crop_image_folder = "crop_images"
```

```python
print("\n⏳ Loading MobileNetV2 model...")
model = MobileNetV2(weights="imagenet", include_top=False, pooling="avg")
print("✅ Model loaded.")
```

```python
print("\n📊 Precomputing embeddings for reference images...")
reference_embeddings = {}

for root, dirs, files in os.walk(crop_image_folder):
    for img_file in files:
        if img_file.lower().endswith((".jpg", ".jpeg", ".png")):
            img_path = os.path.join(root, img_file)
            try:
                img = image.load_img(img_path, target_size=(224, 224))
                x = image.img_to_array(img)
                x = np.expand_dims(x, axis=0)
                x = preprocess_input(x)
                embedding = model.predict(x)
                crop_name = os.path.splitext(img_file)[0]
                reference_embeddings[crop_name] = embedding
                print(f"✅ Processed: {crop_name}")
            except Exception as e:
                print(f"⚠️ Skipped {img_file}: {e}")

print("✅ Embeddings computed for all reference crops.")
```

```python
import importlib
import google.colab.files
files = google.colab.files  # reset files

print("\n📷 Upload a test crop image:")
uploaded_img = files.upload()
test_img_path = list(uploaded_img.keys())[0]
```

```python
img = image.load_img(test_img_path, target_size=(224, 224))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
test_embedding = model.predict(x)
```

```python
max_similarity = -1
best_match = None
for crop, emb in reference_embeddings.items():
    similarity = cosine_similarity(test_embedding, emb)[0][0]
    if similarity > max_similarity:
        max_similarity = similarity
        best_match = crop

print(f"\n🔎 AI Detected Crop: {best_match} (Similarity: {max_similarity*100:.2f}%)")
```

```python
match = df[df["Crop Name"].str.contains(best_match, case=False, na=False)]
if not match.empty:
    print("\n🌱 Crop Found in Dataset:", match["Crop Name"].values[0])
    print("📍 Region:", match["Indigenous Region"].values[0])
    print("🧪 Phytochemicals:", match["Phytochemicals (Major)"].values[0])
    print("💊 Anti-Cancer Evidence:", match["Anti-Cancer Evidence"].values[0])
    print("⚡ Underutilized:", match["Underutilized Status"].values[0])
else:
    print("\n⚠️ Crop not found in dataset.")
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
# Display the uploaded test image and the predicted crop
plt.figure(figsize=(4,4))
plt.imshow(image.load_img(test_img_path))
plt.axis('off')
plt.title(f"AI Detected: {best_match}\nSimilarity: {max_similarity*100:.2f}%")
plt.show()
```

```python
# Compute similarity to every reference embedding and plot top-K
all_scores = []
for crop, emb in reference_embeddings.items():
    sim = cosine_similarity(test_embedding, emb)[0][0]
    all_scores.append((crop, sim))

# sort and keep top 8 (adjust K as you like)
all_scores.sort(key=lambda x: x[1], reverse=True)
top_k = all_scores[:8]

labels = [c for c,_ in top_k]
scores = [s for _,s in top_k]

plt.figure(figsize=(8,4))
plt.bar(labels, scores)
plt.xticks(rotation=30, ha='right')
plt.ylabel("Cosine similarity")
plt.title("Top similar reference images")
plt.ylim(0, 1.0)
plt.tight_layout()
plt.show()
```

```python
# Stack reference embeddings for PCA
ref_names = []
ref_vecs = []

for name, emb in reference_embeddings.items():
    # emb has shape (1, 1280) for MobileNetV2 (avg pooled); squeeze to (1280,)
    ref_names.append(name)
    ref_vecs.append(emb.squeeze())

ref_vecs = np.vstack(ref_vecs)  # shape: (N, 1280)

# Fit PCA on reference vectors and transform both refs + test image
pca = PCA(n_components=2, random_state=0)
ref_2d = pca.fit_transform(ref_vecs)
test_2d = pca.transform(test_embedding)

plt.figure(figsize=(6,6))
plt.scatter(ref_2d[:,0], ref_2d[:,1], s=40)
for i, name in enumerate(ref_names):
    plt.annotate(name, (ref_2d[i,0], ref_2d[i,1]), fontsize=8, xytext=(3,3), textcoords='offset points')

# highlight the test point
plt.scatter(test_2d[0,0], test_2d[0,1], s=120, marker='*')
plt.annotate("TEST", (test_2d[0,0], test_2d[0,1]), fontsize=10, weight='bold', xytext=(5,5), textcoords='offset points')

plt.title("Reference embedding map (PCA)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.tight_layout()
plt.show()
```

```python
# Try to locate the matched reference image file and display side-by-side
matched_path = None
for root, dirs, files_ in os.walk(crop_image_folder):
    for f in files_:
        if f.lower().endswith((".jpg", ".jpeg", ".png")):
            if os.path.splitext(f)[0].lower() == best_match.lower():
                matched_path = os.path.join(root, f)
                break
    if matched_path:
        break

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image.load_img(test_img_path))
plt.axis('off')
plt.title("Test image")

plt.subplot(1,2,2)
if matched_path and os.path.exists(matched_path):
    plt.imshow(image.load_img(matched_path))
    plt.title(f"Best match: {best_match}")
else:
    plt.text(0.5, 0.5, "Matched ref image not found", ha='center', va='center')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Build full similarity list for the test image
all_scores = []
for crop, emb in reference_embeddings.items():
    sim = cosine_similarity(test_embedding, emb)[0][0]
    all_scores.append((crop, sim))
all_scores.sort(key=lambda x: x[1], reverse=True)

scores = [s for _, s in all_scores]

# Histogram
plt.figure(figsize=(6,4))
plt.hist(scores, bins=20)
plt.title("Similarity distribution (test vs. all references)")
plt.xlabel("Cosine similarity")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# ECDF
scores_sorted = sorted(scores)
y = [(i+1)/len(scores_sorted) for i in range(len(scores_sorted))]
plt.figure(figsize=(6,4))
plt.plot(scores_sorted, y)
plt.title("ECDF of similarities")
plt.xlabel("Cosine similarity")
plt.ylabel("Proportion ≤ x")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Build a matrix of reference embeddings
ref_names = []
ref_vecs = []
for name, emb in reference_embeddings.items():
    ref_names.append(name)
    ref_vecs.append(emb.squeeze())
ref_vecs = np.vstack(ref_vecs)

MAXN = 50  # sample to avoid gigantic heatmaps
idx = list(range(min(MAXN, len(ref_names))))
names_s = [ref_names[i] for i in idx]
vecs_s = ref_vecs[idx]

# pairwise cosine similarity
# normalize vectors
norms = np.linalg.norm(vecs_s, axis=1, keepdims=True) + 1e-8
vecs_n = vecs_s / norms
sim_mat = vecs_n @ vecs_n.T  # (N x N)

plt.figure(figsize=(7,6))
plt.imshow(sim_mat, interpolation='nearest')
plt.title("Reference–Reference similarity heatmap")
plt.colorbar(label="Cosine similarity")
plt.xticks(range(len(names_s)), names_s, rotation=90, fontsize=7)
plt.yticks(range(len(names_s)), names_s, fontsize=7)
plt.tight_layout()
plt.show()
```
